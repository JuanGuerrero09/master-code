## Question 1: 

Calculate the mean value μ and the variance σ2 of the sample for each rainfall duration T.  

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('data_exercise_03.csv')
data.head()
```

```python
# import dataclass
from dataclasses import dataclass

@dataclass
class RainfallEvent:
  duration: str
  mean: float
  std: float
  variance: float

# make year column as index
data = data.set_index('Year')

# iterate by name of columns
rainfall_events = []
for col in data.columns:
    mean = data[col].mean()
    std = data[col].std()
    variance = std ** 2
    duration = col
    rainfall_event = RainfallEvent(duration,mean, std, variance)
    rainfall_events.append(rainfall_event)

for event in rainfall_events:
    print(event)
```

## Question 2
Estimate the parameters u and α of the Gumbel distribution for each rainfall duration T using  the methods of moments with the constant γe = 0.5772.  

```python
def get_gumbel_params(mean, std):
    """Calculate Gumbel distribution parameters from mean and standard deviation."""
    gamma = 0.5772  # Euler-Mascheroni constant
    beta = std * np.sqrt(6) / np.pi # beta is alpha
    mu = mean - beta * gamma #mu is u
    return mu, beta

@dataclass
class RainfallGumbelParams:
  duration: str
  alpha: float
  u: float

gumbel_params_list = []
for event in rainfall_events:
    mu, beta = get_gumbel_params(event.mean, event.std)
    gumbel_params = RainfallGumbelParams(event.duration, beta, mu)
    gumbel_params_list.append(gumbel_params)

for params in gumbel_params_list:
    print(params)
```
