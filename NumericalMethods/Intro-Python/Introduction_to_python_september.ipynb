# If statements 

The if statement is a conditional statement in Python used to determine whether a block of code will be executed. If the program finds the condition defined in the if statement true, it will execute the code block inside the if statement.


```python
a = 7
if a<5:
    print('less than five')
```


## Multiple if statements

If-statements can be chained together one after another to create a programmatic flow. For example, the following code block utilizes three different if-statements, each if-statement is followed by an indented code block.


```python
a = 5
if a<0:
    print('is negative')
if a > 0:
    print('is zero')
if a>0:
    print('is positive')
```


## If else statements
An if statement can have an optional else clause. The else statement executes if the condition in the if statement evaluates to False.


```python
var = 7
if type(var)==int:
    print('Input is valid')
else:
    print('Input is invalid')
```


## Elif statements
Elif is the shortened version of else if. It enables you to perform a series of checks to evaluate the conditions of multiple expressions.

```python
price = 8

if price > 100:
    print("price is greater than 100")
elif price == 100:
    print("price is 100")
elif price < 100:
    print("price is less than 100")
```

## Nested if-elif-else conditions
You can nest if statements inside other if statements.

```python
temp = 70
humidity = 100

if temp > 68 and temp <= 82:
    if humidity <= 60:
        print("It's really nice out")
    else:
        print("It's warm, but too humid out there")
elif temp >82:
    print("It's very hot out")
else:
    print("It's cold out there.")
```

# Loops

A loop is an instruction that repeats multiple times as long as some condition is met.<be>
There are mainly two types of loops:

## For Loop
A for loop in Python is used to iterate over a sequence (list, tuple, set, dictionary, and string)<br>

```python
nums = (1, 2, 3, 4)
sum_nums = 0
for num in nums:
    sum_nums += num
print(f'Sum of numbers is {sum_nums}')
```

Note, that this loop directly iterates over elements in the array. You don't need to address the length of the array. <br>
But you can do loops similar to Matlab syntaxis using "range()". **Not recommended to use them** <br>
Examples:

```python
# Loop from 0 to 4
for i in range(5):
    print(i)
```

```python
# Loop from 1 to 4
for i in range(2, 10,2):
    print(i)
```

```python
# Loop from 5 to 1, decrementing by 1
for i in range(5, 0, -1):
    print(i)
```

## While Loop
While loops repeat as long as a certain boolean condition is met. For example:


```python
i = 0
while i < 10:
    print(i, end=" ")
    i += 1
```

This loop also prints numbers 0 to 9, but notice how the condition i < 10 is checked before each iteration, and i is manually incremented within the loop.

## Solution of System of Equations with Numpy and Skipy

In thisscipy.linalg.solve example, we demonstrate how to solve a system of linear equations using Python, leveraging the power of **NumPy** and **Skipy** libraries. 

We define a matrix  $A$ and a vector $\mathbf{b}$, representing a system of linear equations:

$A \mathbf{x} = \mathbf{b}$


Where:
- $A$ is a 3x3 matrix of coefficients,
- $\mathbf{b}$ is the right-hand side vector,
- $\mathbf{x}$ is the unknown vector we want to solve for.

#### System of Equations

The system of equations in our case is:

\begin{aligned}
2x_1 + x_2 + 3x_3 &= 14 \\
x_1 + 3x_2 + 2x_3 &= 13 \\
4x_1 + 2x_2 + x_3 &= 16
\end{aligned}

We can represent this system in matrix form and solve for $\mathbf{x}$ using `solve` function from **Scipy**.

#### Steps:
1. **Define the matrix**  $A$ and vector $\mathbf{b}$.
2. **Use `solve`** to compute the solution for $\mathbf{x}$.
3. **Print output**



```python
import numpy as np
from scipy.linalg import solve

# Step 1: Define a 3x3 matrix A and a vector b
A = np.array([[2, 1, 3], [1, 3, 2], [4, 2, 1]])  # 3x3 matrix
b = np.array([14, 13, 16])  # 3-element vector

# Step 2: Solve the system of equations Ax = b using scipy solve
x = solve(A, b)

# Step 3: Output the solution
print("Solution vector x:", x)
```

Other solvers can be used to resolve the system of equations. For example, Conjugate Gradient and Bi-Congugate gradient 

```python
from scipy.sparse.linalg import cg, bicg

# Step 2.1: Solve the system using Conjugate Gradient (CG)
x_cg, info_cg = cg(A, b,tol=1e-6)

# Step 2.2: Solve the system using Bi-Conjugate Gradient (BiCG)
x_bicg, info_bicg = bicg(A, b,tol=1e-2)

# Step 3: Output the solutions
print("Solution vector x (Conjugate Gradient):", x_cg)
print("Solution vector x (Bi-Conjugate Gradient):", x_bicg)
```

```python
#Check the solution vector
solution = x_bicg
print(A[0][0]*solution[0] + A[0][1]*solution[1] + A[0][2]*solution[2]) 
```

## Pickle file in Python
When working with NumPy, especially in simulation or modeling tasks, it’s often useful not just to perform calculations but also to save and reload data efficiently. This is where Python’s pickle module becomes very handy. Simulations—especially large ones—can be time-consuming to run. If you're tuning parameters, comparing scenarios, or visualizing results later, it’s smart to save intermediate or final data instead of rerunning everything. Storing with .pkl helps with:

Avoiding repeated computation

Debugging and reproducibility

Sharing data between scripts or sessions

The pickle module in Python provides a convenient way to serialize and deserialize Python objects for storage or transfer. The function pickle.dump() is used to save a Python object to a file by converting it into a byte stream—a process known as serialization. Conversely, pickle.load() is used to read the serialized byte stream from a file and reconstruct the original object, a process called deserialization. The protocol parameter in pickle.dump() allows users to specify the version of the pickle format to use; typically, pickle.HIGHEST_PROTOCOL is recommended for efficiency, especially when working with large objects like NumPy arrays. When using pickle, files must be opened in binary mode: 'wb' (write binary) for saving data and 'rb' (read binary) for loading. This ensures that the serialized byte stream is correctly written to and read from disk.

```python
import numpy as np
import matplotlib.pyplot as plt
import pickle

# Parameters
L = 0.2             # Length of rod (m)
n = 200             # Number of spatial points
T0 = 0              # Initial temperature (°C)
TRL = 50            # Left wall temperature (°C)
TRR = 25            # Right wall temperature (°C)
alpha = 0.0001      # Thermal diffusivity (m²/s)
dx = L / n          # Spatial step
t_final = 60        # Final simulation time (s)
dt = 0.001          # Time step
nt = int(t_final / dt)  # Number of time steps

# Grids
x = np.linspace(dx/2, L - dx/2, n)
t = np.linspace(0, t_final, nt)
T = np.ones(n) * T0
T_all = np.zeros((nt, n))  # Store temperatures over time

# Time stepping
for j in range(nt):
    dT = np.zeros(n)
    for i in range(1, n - 1):
        dT[i] = alpha * (T[i + 1] - 2 * T[i] + T[i - 1]) / dx**2
    # Dirichlet BCs
    dT[0] = alpha * (T[1] - 2 * T[0] + TRL) / dx**2
    dT[-1] = alpha * (TRR - 2 * T[-1] + T[-2]) / dx**2

    T += dT * dt
    T_all[j, :] = T

# Save the temperature evolution data using pickle
with open('temperature_data.pkl', 'wb') as f:
    pickle.dump(T_all, f, protocol=pickle.HIGHEST_PROTOCOL)
```

```python
# Load the data back
with open('temperature_data.pkl', 'rb') as f:
    T_all_loaded = pickle.load(f)

# Contour plot
X, T_grid = np.meshgrid(x, t)
plt.figure(figsize=(8, 5))
contour = plt.contourf(X, T_grid, T_all_loaded, 100, cmap='hot')
plt.colorbar(contour)
plt.title("Temperature Evolution in 1D Wall")
plt.xlabel("Position (m)")
plt.ylabel("Time (s)")
plt.show()
```

<b>Velocity components in polar coordinates:<b><br>

$u_r = U_0 \left( 1 - \frac{a^2}{r^2} \right) \cos(\theta)$

$u_\theta = - U_0 \left( 1 + \frac{a^2}{r^2} \right) \sin(\theta)$

$u_y = V_r \sin(\theta) + V_\theta \cos(\theta)$

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
a = 1       # radius of the cylinder
U0 = 1      # free-stream velocity

# Create mesh grid
x = np.linspace(-5, 5, 200)
y = np.linspace(-3, 3, 120)
X, Y = np.meshgrid(x, y)

# Convert Cartesian to polar coordinates
r = np.sqrt(X**2 + Y**2)
theta = np.arctan2(Y, X)

# Avoid division by zero near the center
r = np.where(r < a, a, r)

# Velocity components in polar coordinates
u_r = U0 * (1 - (a**2)/(r**2)) * np.cos(theta)
u_theta = -U0 * (1 + (a**2)/(r**2)) * np.sin(theta)

# Convert to Cartesian velocity components
u_x = u_r * np.cos(theta) - u_theta * np.sin(theta)
u_y = u_r * np.sin(theta) + u_theta * np.cos(theta)

# Mask the cylinder region
mask = r <= a
u_x[mask] = np.nan
u_y[mask] = np.nan

# Plot vector field
plt.figure(figsize=(10, 6))
plt.quiver(X, Y, u_x, u_y, scale=30, pivot='middle')
circle = plt.Circle((0, 0), a, color='green', fill=False, linewidth=0.5)
plt.gca().add_patch(circle)
plt.title("Velocity Field over a Cylinder")
plt.xlabel("x")
plt.ylabel("y")
plt.axis('equal')
plt.grid(True)
plt.xlim(-5, 5)
plt.ylim(-3, 3)
plt.show()
```

## Aliasing

Aliasing occurs when two or more variables refer to the same NumPy array object in memory. Since they are simply different names for the same underlying data, any change made through one variable will also affect the other.

In Python, variables that hold arrays do not actually store the data themselves — they store references to the memory location where the array is stored. So, if you assign one array to another variable using b = a, both variables will point to the same array.

To create a truly independent copy of the array — one that does not reflect changes made to the original — use np.copy(a) or a.copy().



```python
import numpy as np

# Create an array
a = np.array([1, 2, 3, 4, 5])

# Aliasing: b points to the same array as a
b = a

# Modify one element through 'a'
a[0] = 99

# Print both arrays
print("Array a:", a)
print("Array b:", b)
```

### Debugging of 1D diffusion equation
While coding there are some common errors that occur which are explained by 1D diffusion equation given below . First and foremost waz to start checking error is read the error message carefully , it's usually your best clue. Each of the stage you can give a print statement and debug to see if its behaving erratically.

```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm

# Parameters
L = 10.0        # Length of the domain
T = 1           # Total time
Nx = 100        # Number of spatial points
Nt = 10      # Number of time steps
D = 0.05        # Diffusion coefficient
c_coef = 10     

#Discretization
dx = L / Nx               
dt = T / Nt
r = D * dt / dx**2


# Grid
x = np.linspace(0, L, Nx)
t = np.linspace(0, T, Nt+1)
phi = np.zeros(Nx)
phi[Nx // 2] = 10           # Initial condition

# Store 


solution = np.zeros((Nt+1, Nx))
solution[0, :] = phi.copy()

# Time stepping
for n in tqdm(range(1, Nt+1)):
    A = np.zeros((Nx, Nx))
    b = np.zeros(Nx)

    for i in range(Nx):
        A[i, i] = 1 - 2 * r 
        b[i] = phi[i]

        if i > 0:
            A[i, i-1] = r   
        else:
             b[i] = r * c_coef  

        if i < Nx - 1:
            A[i, i+1] = - r   
        else:
            b[i] +=  c_coef


    phi = np.linalg.solve(A, b)
    solution[n, :] = phi.copy()

# Plotting (might produce oscillations or unstable output)
X, Y = np.meshgrid(x, t)
plt.figure(figsize=(8, 5))
cp = plt.contourf(X, Y, solution, levels=50)
plt.colorbar(cp, label='phi(x, t)')
plt.xlabel('x')
plt.ylabel('Time')
plt.title('1D Diffusion')
plt.show()
```

In diffusion equations, stability and accuracy are tied closely to the space and time discretization.Even though you're using the implicit Euler scheme (which is unconditionally stable), a very large r leads to a very stiff system the matrix becomes ill-conditioned, and the numerical solution can have low accuracy
Keeping Nt = 10 means this will effect dt ,$dt= \frac{T}{Nt} =1.0$ which inturn effects the stability criteria.
So use  more time steps for better resolution like Nt = 1000


```python
#Discretization
dx = L / Nx               
dt = T / Nt
```

When we discretize a domain into Nx points, the number of intervals is Nx - 1.
dx is the distance between points, so it should be:<br>
$ 
dx = \frac{L}{Nx - 1}
$<br>
Using dx = L / Nx underestimates the spacing, so all diffusion terms become artificially larger.

```python
r = D * dt / dx**2
```

The **CFL condition** (Courant–Friedrichs–Lewy condition) for **stability in explicit schemes** is given by:

$$
r = D \cdot \frac{\Delta t}{\Delta x^2} \leq 0.5
$$
High r values blow up the matrix entries (1 + 2r, -r), so:

- The system matrix A becomes numerically unstable
- Solving A x = b leads to large values in x (i.e., phi)
- Overflow in values → NaNs, Infs or meaningless solutions.Even if the code  "runs," the result is wrong.<br>
You could solve this by :
- The  most direct and common fix is reduce dt $\Delta t = 0.5 \cdot \frac{\Delta x^2}{D}$
- You could reduce spatial resolution by increasing dx, which would reduce r. But this hurts the accuracy of your spatial solution

```python
  for i in range(Nx):
        A[i, i] = 1 - 2 * r 
        b[i] = phi[i]

        if i > 0:
            A[i, i-1] = r 
```

In implicit Euler, the finite difference discretization of the 1D diffusion equation is:<br>

$$
\frac{\phi_i^{n+1} - \phi_i^n}{\Delta t} = D \cdot \frac{\phi_{i+1}^{n+1} - 2\phi_i^{n+1} + \phi_{i-1}^{n+1}}{\Delta x^2}
$$

This can be rearranged into a linear system of equations:<br>

$$
(1 + 2r) \cdot \phi_i^{n+1} - r \cdot \phi_{i-1}^{n+1} - r \cdot \phi_{i+1}^{n+1} = \phi_i^n
$$

where

$$
r = D \cdot \frac{\Delta t}{\Delta x^2}
$$


The matrix $A$ in the linear system $A \phi^{n+1} = b$ should have:<br>

- Positive main diagonal: $1 + 2r$
- Negative off-diagonals: $-r$

This results in a **tridiagonal system** with the following structure:<br>
[ 1+2r -r 0 ... 0 ] [ -r 1+2r -r ... 0 ] [ 0 -r 1+2r ... 0 ] [ ... ... ... ] [ 0 ... -r 1+2r -r ] [ 0 ... 0 -r 1+2r ]

Using incorrect signs in the matrix (e.g., positive off-diagonals or a negative main diagonal) will invert the direction of diffusion, resulting in **numerical anti-diffusion**.

This causes the solution to become unstable, producing **growing oscillations** or even leading to **blow-up**, instead of the expected smoothing effect of diffusion.

```python
 for i in range(Nx):
        A[i, i] = 1 - 2 * r 
        b[i] = phi[i]

        if i > 0:
            A[i, i-1] = r   
        else:
             b[i] = r * c_coef  

        if i < Nx - 1:
            A[i, i+1] = - r   
        else:
            b[i] +=  c_coef
```

The construction of the b vector is crucial, as it directly impacts the accuracy and stability of the solution. In this case, since we're using periodic boundary conditions, it's essential to apply them correctly.

For periodic boundaries, the function wraps around:
**Left Boundary** \((x=0)\):
$$
\phi_{0,j}^{t+1} = \phi_{\text{max}, j}
$$

**Right Boundary**:
$$
\phi_{\text{max}, j} = \phi_{0,j}
$$
A common mistake when implementing these conditions is in the handling of the b vector, especially regarding how boundary terms are incorporated. Since these boundary values are known through the periodic condition, they must be moved to the right-hand side of the equation system. This is where the sign becomes important.Using += in this context can lead to errors because it implies that you’re adding extra contributions rather than correctly isolating the known terms. Instead, you should ensure that the known boundary terms are carefully and explicitly placed into b, reflecting the mathematical reorganization of the equation. In short: the b vector should only contain terms that are not implicitly handled by the matrix A, and improper sign usage or redundant additions can distort the final solution.

```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm

# Parameters
L = 10.0        # Length of the domain
T = 1          # Total time
Nx = 100        # Number of spatial points
Nt = 1000       # Number of time steps
D = 0.05       # Diffusion coefficient
c_coef = 10
# Discretization
dx = L / (Nx - 1)  #✅ Fixed to get correct spacing
dt = T / Nt
r = D * dt / dx**2 #✅ CFL parameter corrected by adjusting dt
print(f"CFL (r): {r:.4f}")

# Grid
x = np.linspace(0, L, Nx)
t = np.linspace(0, T, Nt+1)
phi = np.zeros(Nx)
phi[Nx // 2] = 10  # Initial condition

# Store solution for contour plot
solution = np.zeros((Nt+1, Nx))
solution[0, :] = phi.copy()

# Time stepping
for n in tqdm(range(1, Nt+1)):
    A = np.zeros((Nx, Nx))
    b = np.zeros(Nx)

    for i in range(Nx):
        A[i, i] = 1 + 2 * r  #✅Wrong sign in diagonal corrected
        b[i] = phi[i]

        if i > 0:
            A[i, i-1] = -r
        else:
            # Periodic BC at left: φ[0] = φ[1]
            b[i] += (r * c_coef) # combine effect at left boundary

        if i < Nx - 1:
            A[i, i+1] += -r
        else:
            # Perriodic BC at right: φ[-1] = φ[-2]
            b[i] += (r * c_coef) #✅Boundary condition errors fixed

    # Solve the system
    phi = np.linalg.solve(A, b)

    # Save solution
    solution[n, :] = phi.copy()

# Contour plot
X, Y = np.meshgrid(x, t)
plt.figure(figsize=(8, 5))
cp = plt.contourf(X, Y, solution, levels=50)
plt.colorbar(cp, label='phi(x, t)')
plt.xlabel('x')
plt.ylabel('Time')
plt.title('1D Diffusion ')
plt.show()
```

**1. CPU Registers**


Registers are the fastest memory units inside the CPU. They store the operands and results of computations and CPU instructions.
When Python code is executed, the low-level CPU instructions generated by the interpreter make use of registers for intermediate steps.
Though Python does not expose registers, the compiled machine-level instructions make full use of them during arithmetic operations, function returns, etc.

**2. Cache Memory**

Cache memory sits between the CPU and RAM. It is faster than RAM but slower than registers. Cache stores recently accessed memory locations from RAM to reduce latency.Helps the CPU avoid fetching data from RAM repeatedly.

Example :

If a variable like x = 5 is used repeatedly, and if it's frequently accessed, its memory location or the result of operations involving it may be cached to speed up performance.

**3. RAM (Random Access Memory)**

RAM is volatile memory that stores the active data and code during the execution of programs.When a Python program runs, all its functions, variables, and data structures reside in RAM.RAM access is significantly faster than disk, but slower than cache.

Divisions within RAM:

a. Stack Memory
Used to manage function calls and local variables.When a function is called, a new stack frame is created with its parameters and temporary variables.
The stack automatically allocates and deallocates memory as functions are called and return.

b. Heap Memory
Used to store dynamically allocated variables like Python objects, lists, dictionaries, arrays, etc.Python stores all objects on the heap and uses references (pointers) to access them.

Memory management here is handled by Python’s garbage collector.

Examples:

x = 5: The integer object 5 is stored in heap memory.

np.array(10**30): This requests a massive amount of heap memory, which exceeds available RAM and results in an Out of Memory (OOM) error.

**4. Storage (Disk – SSD/HDD)**

Storage devices (such as SSDs or HDDs) provide long-term, persistent memory.Unlike RAM, data stored here remains intact after shutdown.

Data from RAM is written to storage via files in python.

This is used for:

Saving scripts

Reading/writing datasets

Persisting variables using serialization modules like pickle, json, or joblib.

Example:

In the diagram, x = 5 is stored in RAM (heap memory). When pickle.dump(x, f) is called, the following happens:

The object x is serialized into a byte stream using the pickle module.

This byte stream is written to a file x.pkl on the storage device.
The result is a binary file like:
b'\x80\x04K\x05.'
which represents the serialized form of the integer 5.

```python
import cv2
import os

def create_video_from_images(image_folder, output_video_path, fps=24):
    # Get list of image files
    images = sorted([img for img in os.listdir(image_folder) if img.endswith((".png", ".jpg", ".jpeg"))])
    
    if not images:
        raise ValueError("No images found in the folder.")

    # Read the first image to get frame dimensions
    first_image_path = os.path.join(image_folder, images[0])
    frame = cv2.imread(first_image_path)
    height, width, _ = frame.shape

    # Define video codec and create VideoWriter object
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # For .mp4 output
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

    for image in images:
        img_path = os.path.join(image_folder, image)
        frame = cv2.imread(img_path)
        if frame is not None:
            resized = cv2.resize(frame, (width, height))
            out.write(resized)
        else:
            print(f"Skipping unreadable image: {img_path}")

    out.release()
    print(f"Video saved to {output_video_path}")

# Example usage
image_folder = "C:\\Users\\Sehrish\\Documents\\frame"
output_video_path = "output_video.mp4"
create_video_from_images(image_folder, output_video_path, fps=10)
```
